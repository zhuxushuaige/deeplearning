```python
import torch
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f'Using {device} device')
torch.manual_seed(777)
if device == 'cuda':
    torch.cuda.manual_seed_all(777)
X = torch.FloatTensor([[0,0],[0,1],[1,0],[1,1]]).to(device)
Y = torch.FloatTensor([[0],[1],[1],[0]]).to(device)
linear = torch.nn.Linear(2,1).to(device)
sigmoid = torch.nn.Sigmoid().to(device)
criterion = torch.nn.BCELoss().to(device)#Binary Cross Entropy
model = torch.nn.Sequential(linear,sigmoid).to(device)
optimizer = torch.optim.SGD(model.parameters(),lr=0.1)

for step in range(10001):
    hypothesis = model(X)
    optimizer.zero_grad()
    cost = criterion(hypothesis,Y)
    cost.backward()
    optimizer.step()
    if step % 100 ==0:
        print(step,cost.item())
        
with torch.no_grad():
    hypothesis = model(X)
    predicted = (hypothesis > 0.5).float()
    accuracy = (predicted == Y).float().mean()
    print('\nHypothesis: ',hypothesis.cpu().numpy())
     
```

```python
import torch
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f'Using {device} device')
torch.manual_seed(777)
if device == 'cuda':
    torch.cuda.manual_seed_all(777)
x = torch.FloatTensor([[0,0],[0,1],[1,0],[1,1]]).to(device)
y = torch.FloatTensor([[0],[1],[1],[0]]).to(device)
linear1 = torch.nn.Linear(2,2,bias=True).to(device)
linear2 = torch.nn.Linear(2,1,bias=True).to(device)
sigmoid = torch.nn.Sigmoid().to(device)
criterion = torch.nn.BCELoss().to(device)#Binary Cross Entropy
model = torch.nn.Sequential(linear1,sigmoid,linear2,sigmoid).to(device)
optimizer = torch.optim.SGD(model.parameters(),lr=1)
for step in range(10001):
    hypothesis = model(x)
    optimizer.zero_grad()
    cost = criterion(hypothesis,y)
    cost.backward()
    optimizer.step()
    if step % 100 ==0:
        print(step,cost.item())
        
with torch.no_grad():
    hypothesis = model(x)
    predicted = (hypothesis > 0.5).float()
    accuracy = (predicted == y).float().mean()
    print('\nHypothesis: ',hypothesis.cpu().numpy())    
    print('Accuracy: ',accuracy.item())
```

```python
import torch
import torchvision.transforms as transforms
import torchvision.datasets as dsets

device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f'Using {device} device')
torch.manual_seed(777)
if device == 'cuda':
    torch.cuda.manual_seed_all(777)
learning_rate = 0.5
batch_size = 10
mnist_train = dsets.MNIST(root='MNIST_data/',
                          train=True,   
                          transform=transforms.ToTensor(),
                          download=True)
mnist_test = dsets.MNIST(root='MNIST_data/',
                         train=False,
                         transform=transforms.ToTensor(),
                         download=True)
data_loader = torch.utils.data.DataLoader(dataset=mnist_train,
                                          batch_size=batch_size,
                                          shuffle=True,
                                          drop_last=True)   
w1 = torch.zeros(784,30,device=device)
b1 = torch.zeros(30,device=device)
w2 = torch.zeros(30,10,device=device)    
b2 = torch.zeros(10,device=device)
torch.nn.init.normal_(w1)
torch.nn.init.normal_(b1)   
torch.nn.init.normal_(w2)
torch.nn.init.normal_(b2)

def sigmoid(x):
    return 1 / (1 + torch.exp(-x))

def sigmoid_prime(x):
    return sigmoid(x) * (1 - sigmoid(x))
X_test = mnist_test.test_data.view(-1,28*28).float().to(device)
Y_test = mnist_test.test_labels.to(device)
i = 0
while i!=20:
    i+=1
    for X, Y in data_loader:
        X = X.view(-1,28*28).to(device)
        Y = torch.zeros(batch_size,10).scatter_(1,Y.unsqueeze(1),1).to(device)
        l1 = torch.add(torch.matmul(X,w1),b1)
        a1 = sigmoid(l1)
        l2 = torch.add(torch.matmul(a1,w2),b2)
        y_pred = sigmoid(l2)
        
        diff = y_pred - Y
        
        d_l2 = diff * sigmoid_prime(l2)
        d_b2 = d_l2
        d_w2 = torch.matmul(torch.transpose(a1,0,1),d_l2)
        
        d_a1 = torch.matmul(d_l2,torch.transpose(w2,0,1))
        d_l1 = d_a1 * sigmoid_prime(l1)
        d_b1 = d_l1
        d_w1 = torch.matmul(torch.transpose(X,0,1),d_l1)
        
        w1 -= learning_rate * d_w1
        b1 -= learning_rate * torch.mean(d_b1,0)
        w2 -= learning_rate * d_w2
        b2 -= learning_rate * torch.mean(d_b2,0)
        
        if i % 1 == 0:
            l1 = torch.add(torch.matmul(X_test, w1), b1)
            a1 = sigmoid(l1)
            l2 = torch.add(torch.matmul(a1, w2), b2)
            y_pred = sigmoid(l2)
            acct_mat = torch.argmax(y_pred, 1) == Y_test
            acct_res = acct_mat.sum()
            print(acct_res.item())
            
        if i == 20:
            break
        
        
```
